# Task 1

```python
score = int(input("Enter your score: "))
if score > 90:
    print("A")
elif score > 80:
    print("B")
elif score > 70:
    print("C")
elif score > 60:
    print("D")
elif score > 50:
    print("E") 
else:
    print("F")
```

# Task 2

```python
n = int(input("Enter a number: "))
sum = 0
divisible = 0
for i in range(1, n + 1):
    number = i
    if number % 2 == 0:
        sum += number
    if number % 3 == 0 and number % 5 == 0:
        print("3 5")
        divisible += 1
    elif number % 3 == 0:
        print("3")
        divisible += 1
    elif number % 5 == 0:
        print("5")
    else:
        print(number)
print("Sum of even numbers:", sum)
print("Count of divisible numbers by 3:", divisible)
```
